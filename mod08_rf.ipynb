### Feature engineering

```python
def feature_engineering(df=heart):
    df['Male'] = (df.Sex == 'Male').astype(int)
    df['HealthyDiet'] = (df.Diet == 'Healthy').astype(int)
    df['UnhealthyDiet'] = (df.Diet == 'Unhealthy').astype(int)
    df[['BP1','BP2']] = df['Blood Pressure'].str.split('/', n=1, expand=True).astype(int)

feature_engineering()
```

### Create a basic random forest model on the training set.

```python
rf_basic = RFC(max_features=len(indep_vars),
               criterion='log_loss',
               n_estimators=50,
               max_depth=10,
               min_samples_leaf=1,
               bootstrap=True,
               oob_score=True,
               random_state=seed)
rf_basic.fit(X_train, y_train)
```

### We'll use this function to quickly calculate the log loss of a model.

```python
def log_loss(y_true, y_prob, eps=1e-15):
    y_true = np.array(y_true)
    y_prob = np.array(y_prob)
    
    # avoid log(0)
    y_prob = np.clip(y_prob, eps, 1 - eps)
    
    ll = -np.mean(y_true * np.log(y_prob) + (1 - y_true) * np.log(1 - y_prob))
    return ll
```

### Compute the log loss on the training set.

```python
y_hat_basic_train = rf_basic.predict_proba(X_train)[:,1]
print('train log loss:', log_loss(y_train, y_hat_basic_train))
```

### Calculate the out-of-bag error (bootstrapping) on training data to estimate the log loss on the test set.

```python
y_hat_basic_oob = rf_basic.oob_decision_function_[:,1]
print('oob log loss:  ', log_loss(y_train, y_hat_basic_oob))
```

### Compute the log loss on the test set.

```python
y_hat_basic_test = rf_basic.predict_proba(X_test)[:,1]
print('test log loss: ', log_loss(y_test, y_hat_basic_test))
```

### Build an improved random forest model on this same set of data. 

You will be scored according to how well it performs on an unseen holdout set.

```python
# Documentation: https://scikit-learn.org/dev/modules/generated/sklearn.ensemble.RandomForestRegressor.html

rf_better = RFC(
    n_estimators=200,
    max_features='sqrt',
    max_depth=6,
    min_samples_leaf=5,
    bootstrap=True,
    oob_score=True,
    random_state=seed
)
rf_better.fit(X_train, y_train)
```

## Discussion Questions

### (1) Explain what "value" means in the decision tree plots.

In a decision tree plot, the “value” at a node represents the number of observations for each class that reach that node. The value helps determine the predicted class at that node, since the model will assign the class with the majority of observations. As we move down the tree, the goal is to create nodes where most observations belong to a single class, making predictions more accurate.

### (2) How is “log loss” different from the log likelihood we defined in class?

Log likelihood measures how well a model explains the observed data by calculating the probability of the data given the model. A higher log likelihood means the model fits the data better. But log loss is based on log likelihood but is used as a metric for error. It takes the negative average of the log likelihood values, so lower log loss means better model performance. So log likelihood is something we want to maximize, while log loss is something we want to minimize when evaluating predictions.

### (3) Explain how we know the basic model is overfit before attempting to build any other models.

We can tell the basic model is overfit by comparing its performance on the training data to the test data. If the training log loss is much lower than the test log loss, it means the model is doing well on the data it was trained on but not as well on new, unseen data. This shows that the model has learned patterns that are too specific to the training data, rather than general patterns that apply more broadly. Meaning it does not generalize well, which is the main sign of overfitting.
